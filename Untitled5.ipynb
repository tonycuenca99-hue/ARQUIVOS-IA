```python
# Transforma em dicionário "nome -> valor"
params = {}

for _, row in parametros_df.iterrows():
    nome = str(row[0]).strip()
    valor = row[1]

    if pd.isna(valor):
        continue

    # Se for string com vírgula, converte
    if isinstance(valor, str):
        try:
            valor_num = float(valor.replace(",", "."))
        except:
            continue
    else:
        valor_num = float(valor)

    params[nome] = valor_num

print("Parâmetros lidos do Excel:\n")
for k, v in params.items():
    print(f"{k}: {v}")
```

```python
rho         = params.get("ρ (kg/m³)", 1000.0)
g           = params.get("g (m/s²)", 9.81)
eta_turbina = params.get("η_turbina", 0.90)
eta_gerador = params.get("η_gerador (valor inicial)", 0.98)
eta_global  = params.get("η_global", eta_turbina * eta_gerador)
H_projeto   = params.get("H_projeto", 40.0)
H_efetivo   = params.get("H_efetivo", H_projeto * (1 - params.get("Perda de carga estimada", 0.05)))
cota_cidade = params.get("Cota da cidade", 312.0)

print("\nParâmetros processados:")
print("rho =", rho)
print("g =", g)
print("eta_turbina =", eta_turbina)
print("eta_gerador =", eta_gerador)
print("eta_global =", eta_global)
print("H_projeto =", H_projeto)
print("H_efetivo =", H_efetivo)
print("Cota da cidade =", cota_cidade)
```

```python
import pandas as pd
import numpy as np

file_path = "DADOS_12.xlsx"

# === 1) Ler a aba COTA AREA ===
cota_area = pd.read_excel(file_path, sheet_name="COTA AREA")

# Garante nomes de colunas consistentes (ajuste se seu nome estiver um pouco diferente)
cota_col = "Cota (m)"
area_km2_col = "Area Inundada (km2)"

# Mantém só as colunas importantes
cota_area = cota_area[[cota_col, area_km2_col]].copy()

# Remove linhas totalmente vazias
cota_area = cota_area.dropna(how="all")

# Converte cota para float
cota_area[cota_col] = (
    cota_area[cota_col]
    .astype(str)
    .str.replace(",", ".", regex=False)
    .astype(float)
)

# Converte área de km² para float
cota_area[area_km2_col] = (
    cota_area[area_km2_col]
    .astype(str)
    .str.replace(",", ".", regex=False)
    .astype(float)
)

# Ordena por cota, se ainda não estiver
cota_area = cota_area.sort_values(by=cota_col).reset_index(drop=True)

# === 2) Calcular Área em m² ===
cota_area["Area_m2"] = cota_area[area_km2_col] * 1e6

# === 3) Calcular Volume Incremental (m³) usando método dos trapézios ===
# Para a primeira linha, volume incremental = 0
vol_inc = [0.0]

for i in range(1, len(cota_area)):
    h_atual = cota_area.loc[i, cota_col]
    h_ant   = cota_area.loc[i-1, cota_col]
    A_atual = cota_area.loc[i, "Area_m2"]
    A_ant   = cota_area.loc[i-1, "Area_m2"]

    dh = h_atual - h_ant  # diferença de cota (m)
    V_inc_i = (A_atual + A_ant) / 2.0 * dh  # m² * m = m³

    vol_inc.append(V_inc_i)

cota_area["Volume_Incremental_m3"] = vol_inc

# === 4) Volume Acumulado (m³) ===
cota_area["Volume_Acumulado_m3"] = cota_area["Volume_Incremental_m3"].cumsum()

print(cota_area)
```

```python
import pandas as pd
import numpy as np

file_path = "DADOS_12.xlsx"

# === 1) Ler a aba COTA AREA ===
cota_area = pd.read_excel(file_path, sheet_name="COTA AREA")

# Ajuste os nomes de colunas conforme estão na sua planilha
cota_col = "Cota (m)"
area_km2_col = "Area Inundada (km2)"   # ou "Área Inundada (km²)" se for o caso

cota_area = cota_area[[cota_col, area_km2_col]].copy()
cota_area = cota_area.dropna(how="all")

# Converte tipos
cota_area[cota_col] = (
    cota_area[cota_col]
    .astype(str).str.replace(",", ".", regex=False).astype(float)
)

cota_area[area_km2_col] = (
    cota_area[area_km2_col]
    .astype(str).str.replace(",", ".", regex=False).astype(float)
)

# Ordena por cota
cota_area = cota_area.sort_values(by=cota_col).reset_index(drop=True)

# Área em m²
cota_area["Area_m2"] = cota_area[area_km2_col] * 1e6

# Volume incremental (trapézios)
vol_inc = [0.0]
for i in range(1, len(cota_area)):
    h_atual = cota_area.loc[i, cota_col]
    h_ant   = cota_area.loc[i-1, cota_col]
    A_atual = cota_area.loc[i, "Area_m2"]
    A_ant   = cota_area.loc[i-1, "Area_m2"]
    dh = h_atual - h_ant
    V_inc_i = (A_atual + A_ant) / 2.0 * dh
    vol_inc.append(V_inc_i)

cota_area["Volume_Incremental_m3"] = vol_inc
cota_area["Volume_Acumulado_m3"] = cota_area["Volume_Incremental_m3"].cumsum()

print(cota_area)
```

```python
cota_cidade = 312.0

# Vetores para interpolação
cotas = cota_area[cota_col].values
vols  = cota_area["Volume_Acumulado_m3"].values

# Volume total até a cota 312 m (interpola se 312 não estiver exato na tabela)
V_ate_312_m3 = np.interp(cota_cidade, cotas, vols)

# Nível mínimo: a primeira cota da tabela (volume ~0)
cota_min = cotas[0]
V_ate_min_m3 = vols[0]   # deve ser ~0 se na tabela estiver assim

# Volume útil (entre cota_min e 312 m)
V_res_uti_m3 = V_ate_312_m3 - V_ate_min_m3
V_res_uti_hm3 = V_res_uti_m3 / 1e6

print(f"Cota mínima considerada        = {cota_min:.2f} m")
print(f"Cota máxima (cidade)          = {cota_cidade:.2f} m")
print(f"Volume útil até 312 m         ≈ {V_res_uti_hm3:.1f} hm³")
```

```python
# Se ainda não tiver:
dados["Ano"] = dados["Data"].dt.year
dados["Mes"] = dados["Data"].dt.month

import calendar

mensal = (
    dados.groupby(["Ano", "Mes"])["Q_usina_total"]
    .mean()
    .reset_index()
)

mensal["Dias_mes"] = mensal.apply(
    lambda row: calendar.monthrange(int(row["Ano"]), int(row["Mes"]))[1],
    axis=1
)

mensal["V_natural_m3"] = mensal["Q_usina_total"] * mensal["Dias_mes"] * 86400
```

```python
import pandas as pd
import numpy as np
import calendar

file_path = "DADOS_12.xlsx"

# ========== 1) PARÂMETROS (aba PARAMETROS) ==========
parametros_df = pd.read_excel(file_path, sheet_name="PARAMETROS", usecols=[0, 1])
parametros_df = parametros_df.dropna(how="all")

params = {}
for _, row in parametros_df.iterrows():
    nome = str(row[0]).strip()
    valor = row[1]
    if pd.isna(valor):
        continue
    if isinstance(valor, str):
        try:
            valor_num = float(valor.replace(",", "."))
        except:
            continue
    else:
        valor_num = float(valor)
    params[nome] = valor_num

rho         = params.get("ρ (kg/m³)", 1000.0)
g           = params.get("g (m/s²)", 9.81)
eta_turbina = params.get("η_turbina", 0.90)
eta_gerador = params.get("η_gerador (valor inicial)", 0.98)
eta_global  = params.get("η_global", eta_turbina * eta_gerador)
H_projeto   = params.get("H_projeto", 40.0)        # queda bruta de projeto (m)
H_efetivo   = params.get("H_efetivo", 38.0)        # queda líquida de projeto (m)

print("Parâmetros:", rho, g, eta_global, H_projeto, H_efetivo)

# ========== 2) CURVA COTA-VAZÃO (aba COTA_VAZAO) ==========
cota_vazao = pd.read_excel(file_path, sheet_name="COTA_VAZAO")
cota_vazao = cota_vazao.rename(columns={
    "COTA": "cota", # Corrigido aqui: de 'Cota (m)' para 'COTA'
    "Vazão (m³/s)": "vazao"
})
cota_vazao = cota_vazao.dropna(subset=["cota", "vazao"])
cota_vazao["cota"] = cota_vazao["cota"].astype(str).str.replace(",", ".", regex=False).astype(float)
cota_vazao["vazao"] = cota_vazao["vazao"].astype(str).str.replace(",", ".", regex=False).astype(float)
cota_vazao = cota_vazao.sort_values(by="cota").reset_index(drop=True)

def cota_por_vazao(vazoes):
    """Retorna cota (m) via interpolação (curva-chave) para vazão(ões) em m³/s."""
    return np.interp(vazoes, cota_vazao["vazao"], cota_vazao["cota"])

# ========== 3) SAZONALIDADE – vazões médias mensais ==========
saz = pd.read_excel(file_path, sheet_name="SAZONALIDADE")

# Ajusta nomes das colunas conforme sua planilha
saz = saz.rename(columns={
    "NUMERO DE MÊS": "Mes_num",
    "NUMERO DE MES": "Mes_num",   # caso escrito sem acento
    "MÉDIA A": "Q_media_m3s",
    "MEDIA A": "Q_media_m3s"
})

saz = saz[["Mes_num", "Q_media_m3s"]].dropna()

# Converte vazão média para float
saz["Q_media_m3s"] = (
    saz["Q_media_m3s"]
    .astype(str).str.replace(",", ".", regex=False).astype(float)
)

# Garante 12 meses ordenados
saz = saz.sort_values("Mes_num").reset_index(drop=True)

# ========== 4) DEFINIÇÃO DE VAZÃO DE PROJETO (Q_projeto = Q50) ==========
# Se você já tiver Q50 calculado em outra parte, pode usar diretamente.
# Aqui vou supor que Q50 já foi calculado e passado como constante:
Q50 = 38.63  # m³/s

Q_projeto = Q50   # vazão máxima turbinada

# ========== 5) TABELA ENERGÉTICA MENSAL (12 LINHAS) ==========

# DATA representativa: dia 15 de um ano "típico" (2000, 2001 tanto faz)
ano_referencia = 2000
datas = [
    pd.to_datetime(f"{ano_referencia}-{int(mes):02d}-15")
    for mes in saz["Mes_num"]
]

df_mes = pd.DataFrame({
    "DATA": datas,
    "Mes_num": saz["Mes_num"],
    "VAZAO (m³/s)": saz["Q_media_m3s"]
})

# Vazão turbinada = min(vazão média, Q_projeto)
df_mes["Q_turb_m3s"] = df_mes["VAZAO (m³/s)"].clip(lower=0, upper=Q_projeto)

# Nível jusante pela curva-chave (função do Q_turb)
df_mes["Nível jusante (m)"] = cota_por_vazao(df_mes["Q_turb_m3s"])

# Nível montante assumido H_projeto acima do jusante (queda bruta de projeto)
df_mes["Nível montante (m)"] = df_mes["Nível jusante (m)"] + H_projeto

# Queda bruta = diferença entre níveis (≈ H_projeto)
df_mes["Queda bruta (m)"] = df_mes["Nível montante (m)"] - df_mes["Nível jusante (m)"]

# Queda líquida = H_efetivo de projeto
df_mes["Queda líquida (m)"] = H_efetivo

# Potência média em cada mês (kW) – fórmula do livro
df_mes["Potência (kW)"] = (
    rho * g * df_mes["Q_turb_m3s"] * df_mes["Queda líquida (m)"] * eta_global / 1000
)

# Número de dias em cada mês (ano_referencia)
df_mes["Dias_mes"] = df_mes["Mes_num"].apply(
    lambda m: calendar.monthrange(ano_referencia, int(m))[1]
)

# Horas no mês
df_mes["horas_mes"] = df_mes["Dias_mes"] * 24

# Energia mensal (MWh)
df_mes["Energia mensal (MWh)"] = df_mes["Potência (kW)"] * df_mes["horas_mes"] / 1000

# Energia anual (MWh) = soma dos 12 meses
Energia_anual = df_mes["Energia mensal (MWh)"].sum()

# Energia média mensal (MWh/mês)
Energia_media_mensal = Energia_anual / 12.0

# Potência média (kW) = Energia_anual / 8760 h
horas_ano = 365 * 24
Potência_media_kW = Energia_anual * 1000 / horas_ano

# Potência instalada (kW) – usando Q_projeto (=Q50) e H_efetivo
P_inst_W  = rho * g * Q_projeto * H_efetivo * eta_global
P_inst_kW = P_inst_W / 1000

# Fator de capacidade (FC)
FC = Potência_media_kW / P_inst_kW

print("\nRESUMO (ano típico):")
print(f"Energia anual (MWh)             ≈ {Energia_anual:.1f}")
print(f"Energia média mensal (MWh/mês) ≈ {Energia_media_mensal:.1f}")
print(f"Potência média (kW)            ≈ {Potência_media_kW:.1f}")
print(f"P_instalada (kW)               ≈ {P_inst_kW:.1f}")
print(f"Fator de capacidade (FC)       ≈ {FC:.3f}")

# ========== 6) REPETIR RESUMOS EM TODAS AS LINHAS (como no Excel) ==========

df_mes["Energia anual (MWh)"]              = Energia_anual
df_mes["Energia média mensal (MWh/mês)"]   = Energia_media_mensal
df_mes["Potência média (kW)"]              = Potência_media_kW
df_mes["P_instalada (kW)"]                 = P_inst_kW
df_mes["Fator de capacidade (FC)"]         = FC

# Ordena colunas no exato formato que você pediu
tabela_final = df_mes[[
    "DATA",
    "VAZAO (m³/s)",
    "Nível montante (m)",
    "Nível jusante (m)",
    "Queda bruta (m)",
    "Queda líquida (m)",
    "Potência (kW)",
    "Energia mensal (MWh)",
    "Energia anual (MWh)",
    "Energia média mensal (MWh/mês)",
    "Potência média (kW)",
    "P_instalada (kW)",
    "Fator de capacidade (FC)"
]]

print("\nTabela final (12 meses):")
print(tabela_final)
```
